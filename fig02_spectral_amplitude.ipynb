# Figure 2: spectral amplitude

Input: `data/fig02_spectral_amplitude.npz`. For each start time it stores the exact-grid Kerr prediction $|A_\mathrm{GR}|$ and the corresponding free-fit samples $|A_\mathrm{GR}(1-\delta A)|$. Amplitudes are dimensionless strain coefficients.

```python
from pathlib import Path
import warnings
warnings.filterwarnings('ignore', 'Wswiglal-redir-stdio')

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / 'data').is_dir():
    raise RuntimeError('Run this notebook from the repository root.')

fig_width = 3 * 246.0 / 72.27
fig_height = fig_width * (np.sqrt(5) - 1.0) / 2.0
mpl.rcParams.update({
    'axes.labelsize': 24,
    'font.family': 'serif',
    'font.serif': 'Computer Modern Roman',
    'font.size': 24,
    'legend.fontsize': 20,
    'xtick.labelsize': 24,
    'ytick.labelsize': 24,
    'axes.grid': True,
    'text.usetex': True,
    'savefig.dpi': 100,
    'lines.markersize': 14,
    'figure.figsize': [fig_width, fig_height],
})

MTSUN_SI = 4.925490947641267e-6
import matplotlib.patches as mpatches

data = np.load(ROOT / 'data/fig02_spectral_amplitude.npz')
labels = list(range(3, 10))
toffset_ms = np.array([1.00, 1.33, 1.66, 2.00, 2.33, 2.66, 3.00])
kerr = [data[f'kerr_{label}mf'] for label in labels]
free = [data[f'free_{label}mf'] for label in labels]
```

```python
spacing = np.min(np.diff(toffset_ms))
width = 0.35 * spacing
offset = 0.22 * spacing
fig, ax = plt.subplots(figsize=(11, 7))
plot_kerr = ax.violinplot(kerr, positions=toffset_ms - offset, widths=width,
                          showmedians=True, showextrema=False)
plot_free = ax.violinplot(free, positions=toffset_ms + offset, widths=width,
                          showmedians=True, showextrema=False)
for body in plot_kerr['bodies']:
    body.set_facecolor('C0'); body.set_alpha(0.6); body.set_edgecolor('k')
for body in plot_free['bodies']:
    body.set_facecolor('C3'); body.set_alpha(0.6); body.set_edgecolor('k')
plot_kerr['cmedians'].set_color('k')
plot_free['cmedians'].set_color('k')

ax.set_yscale('log')
ax.set_xlabel('Time After Merger (ms)')
ax.set_ylabel(r'$\left|\sum_{n,m} A_\mathrm{22n\times22m}\right|$')
ax.set_xticks(toffset_ms)
ax.set_xticklabels([f'{value:g}' for value in toffset_ms])
ax.legend(handles=[
    mpatches.Patch(color='C0', alpha=0.6, label='Kerr'),
    mpatches.Patch(color='C3', alpha=0.6, label='Free Fit')])

ax_top = ax.twiny()
conversion_factor = 67.68112439526381 * MTSUN_SI
ax_top.set_xticks(toffset_ms)
ax_top.set_xticklabels([f'{value * 1e-3 / conversion_factor:.1f}' for value in toffset_ms])
ax_top.set_xlim(ax.get_xlim())
ax_top.set_xlabel(r'Time After Merger ($M_\mathrm{f}$)')
ax_top.grid(False)

output = ROOT / 'figures/spectral_amplitude_wls10modes_gaussian.pdf'
fig.savefig(output, bbox_inches='tight')
plt.show()
```
